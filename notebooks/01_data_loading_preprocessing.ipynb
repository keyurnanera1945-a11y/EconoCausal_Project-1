# 01 - Data Loading & Preprocessing Pipeline

This notebook demonstrates how to execute the modular Data Pipeline created in `src/` for data loading, missing value imputation, causal feature matrix construction ($X, T, Y$), train-test splitting, and summary table generation.


### 1. Import System & Pipeline Modules


```python
import sys
import os
from pathlib import Path
import pandas as pd
import numpy as np

# Ensure root project directory is in Python path
project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.pipeline import run_pipeline, DataPipeline
from src.config import load_config
```

### 2. Configure Pipeline & Data Paths


```python
config_path = project_root / "config" / "config.yaml"
config = load_config(str(config_path))

# Check if dataset file exists in notebooks/ or config raw_data_path
data_file = project_root / "notebooks" / "criteo-uplift-v2.1.csv"
if not data_file.exists():
    data_file = project_root / config.raw_data_path

print("Config loaded successfully.")
print("Target Data File:", data_file)
```

### 3. Run Pipeline End-to-End


```python
# Execute preprocessing pipeline
pipeline = DataPipeline(str(config_path))
results = pipeline.run(data_path=str(data_file) if data_file.exists() else None)
```

### 4. Inspect Feature Matrices & Train/Test Splits


```python
X_train, X_test = results["X_train"], results["X_test"]
T_train, T_test = results["T_train"], results["T_test"]
Y_train, Y_test = results["Y_train"], results["Y_test"]

print("X_train shape:", X_train.shape)
print("X_test shape: ", X_test.shape)
print("T_train distribution:\n", T_train.value_counts(normalize=True))
print("Y_train distribution:\n", Y_train.value_counts(normalize=True))
```
